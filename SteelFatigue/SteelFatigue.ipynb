```python
df.describe() #basic statistical info
#NT is normalizing temperature 
#NT is a "high-heat" reset used to break down big, ugly chunks of grain.
#THT is the "surgical" heat. You want it just high enough to get that Austenite structure, but low enough that you don't grow "fat" grains.
#if the grains will get too large ,cracks will get an easy path to grow,thats why THT is kept lower than NT

#THt-tells how long to keep the sample at THT(through hardening temperature),
#it is important since the transfer of heat from surface to bulk is not an instant or a fast process.

#THQCr-tells the rate at which Sample cools down to room temp from THT

#THT through hardening temperature is the temp required to heat the bulk again to make the sample ready for quenching

#CT stands for carburization temp,which tells the temp at which steel is heated while surrounded by a carbon rich env
#but acc to data more than 75% steels didnt go under carburization and eventually diffusion

#Ct,Dt-mins,while CT,DT -degree celsius

#QmT-after carburization and diffusing ,sample is red hot and loaded with carbon and therefore we quench it to turn the surface into hard case

#acc to data ,most of samples are through hardened(more than 75%) while only a small group went under carburization ,

#TT -Tempering temperature ,it is the temp at which quenched sample is reheated(below THT)
#Acc to data ,it seems high temp tempering is done,at this temp brittle martensite transforms into Tempered Martensite which inc steels's toughness

#Tt-Tempering time-how long the sample stays in tempering furnace;
#acc to data ,a 60 min time is standard,it ensures the internal stresses are relieved uniformly throughout the piece

#TCr-cooling rate after tempering

#reduction ratio-The ratio of the cross-sectional area of the original steel ingot to the final bar
#For example, a ratio of 10 means the original block was squeezed until it was 10 times thinner/longer.

#dA,dB,dC-Non-metallic Inclusions,like slag or oxides trapped inside the steel.crack starts from here
#dA-Sulfide type, dB-Alumina type, dC-Silicate/Oxide Type
# acc to data-dA is much higher (0.047) than other two

# dA, dB, dC: The higher these values are, the more "starting points" there are for a crack.
# in data with a very low Fatigue value (like the min 225),its dA, dB, or dC are likely higher than average.

#std for carbon is way less than Mn which shows that samples with diff diff Mn levels are used while C% was almost same
```

```python
import matplotlib.pyplot as plt

correlation_matrix = x.corr()

plt.figure(figsize=(14, 10))
sns.heatmap(correlation_matrix, annot=True, cmap='coolwarm', fmt='.2f')
plt.title("Feature Correlation Heatmap")
plt.show()
```

### Identifying Highly Correlated Feature Pairs

Extracting feature pairs with high absolute correlation to investigate multicollinearity.

```python
corr_matrix = x.corr()

high_corr = []

for i in range(len(corr_matrix.columns)):
    for j in range(i + 1, len(corr_matrix.columns)):
        corr_value = corr_matrix.iloc[i, j]

        if abs(corr_value) >= 0.85:
            high_corr.append([
                corr_matrix.columns[i],
                corr_matrix.columns[j],
                corr_value
            ])

high_corr = pd.DataFrame(
    high_corr,
    columns=['Feature 1', 'Feature 2', 'Correlation']
)

high_corr = high_corr.sort_values(
    by='Correlation',
    key=abs,
    ascending=False
)

high_corr
```

### Random Forest Regression

Using a nonlinear ensemble model to capture relationships that may not be represented well by linear regression.

```python
from sklearn.ensemble import RandomForestRegressor

rf_model = RandomForestRegressor(
    n_estimators=200,
    random_state=42
)

rf_model.fit(x_train_original, Y_train_original)

rf_pred = rf_model.predict(x_test_original)
```

```python
rf_r2 = r2_score(Y_test_original, rf_pred)
rf_mae = mean_absolute_error(Y_test_original, rf_pred)
rf_rmse = mean_squared_error(Y_test_original, rf_pred) ** 0.5

print("Random Forest R2:", rf_r2)
print("Random Forest MAE:", rf_mae)
print("Random Forest RMSE:", rf_rmse)
```

```python
feature_importance = pd.DataFrame({
    'Feature': x.columns,
    'Importance': rf_model.feature_importances_
})

feature_importance = feature_importance.sort_values(
    by='Importance',
    ascending=False
)

feature_importance
```

```python
plt.figure(figsize=(10, 6))

plt.barh(
    feature_importance['Feature'].head(10)[::-1],
    feature_importance['Importance'].head(10)[::-1]
)

plt.xlabel("Importance")
plt.ylabel("Feature")
plt.title("Top 10 Feature Importances - Random Forest")
plt.show()
```

## Final Results

The Linear Regression model achieved an R² of 0.9750, MAE of 24.87, and RMSE of 32.16.

Random Forest Regression improved the results to an R² of 0.9855, MAE of 18.99, and RMSE of 24.45 on the test set.

Feature correlation analysis revealed substantial multicollinearity among several heat-treatment parameters.

Two physically motivated interaction features were also tested, but they did not improve the Linear Regression performance and were therefore not retained.

The Random Forest feature-importance analysis identified CT, DT, Tt, Cr, QmT, NT, Ct, Dt, TT, and C among the most influential features in the model.
